```python
# DuckDB Tables Explorer
# get "stock_data" table details in dataframe from DuckDB database.
```

```python
import duckdb
import pandas as pd
from src.core_config import DB_PATH

conn = duckdb.connect(str(DB_PATH))

# 1) Column-level schema details
stock_data_schema_df = conn.execute("PRAGMA table_info('stock_data')").fetchdf()

# 2) Optional: quick table summary
stock_data_summary_df = conn.execute("""
    SELECT
        COUNT(*) AS total_rows,
        MIN(trade_date) AS min_trade_date,
        MAX(trade_date) AS max_trade_date
    FROM stock_data
""").fetchdf()

conn.close()

stock_data_schema_df, stock_data_summary_df
```

```python
conn = duckdb.connect(str(DB_PATH))
stock_data_preview_df = conn.execute("""
    SELECT *
    FROM stock_data
    where symbol = 'RELIANCE'
    ORDER BY trade_date DESC, stock_id
""").fetchdf()
conn.close()

stock_data_preview_df
```
